### Preparing and Merging the Datasets

1. **Inspect `PInj` and clean it**: We need to transpose the wide table so that counties become rows, extract the `Personal Income` row, and clean the county names to match the uppercase names in `njCO['COUNTY']`.
2. **Merge**: Combine the spatial dataset and the cleaned income dataset.

```python
import pandas as pd

# Let's inspect the first column name of PInj (it contains the metric descriptions)
metric_col = PInj.columns[0]

# We will extract the 'Personal Income (thousands of dollars)' row (which is index 1)
income_row = PInj[PInj[metric_col].str.strip() == 'Personal Income (thousands of dollars)'].copy()

# Melt the table to convert county columns into rows
income_melted = income_row.melt(
    id_vars=[metric_col],
    var_name='County_Raw',
    value_name='Personal_Income'
)

# Clean county names to match `njCO['COUNTY']` (e.g., "Atlantic County, New Jersey" -> "ATLANTIC")
income_melted['COUNTY'] = (
    income_melted['County_Raw']
    .str.replace(' County, New Jersey', '', case=False, regex=False)
    .str.strip()
    .str.upper()
)

# Drop totals or non-county columns if any
income_cleaned = income_melted[~income_melted['COUNTY'].str.contains('TOTAL', case=False, na=True)]

# View the prepared data
display(income_cleaned.head())
```

```python
# Merge the spatial data with the cleaned personal income data
njCO_merged = njCO.merge(income_cleaned, on='COUNTY', how='left')

# Display the first few rows of the merged GeoDataFrame
display(njCO_merged[['COUNTY', 'geometry', 'Personal_Income']].head())
```

```python
import matplotlib.pyplot as plt
import geopandas as gpd
fig, ax = plt.subplots(1, figsize=(6,10))
njCO_merged.plot(ax=ax,column='Personal_Income',legend=True,cmap='YlOrRd',scheme='natural_breaks',k=5,
         edgecolor='grey',linewidth=2,legend_kwds= {"fmt": "{:,.0f}",
        'loc':'lower right','title_fontsize':'medium','fontsize':'small','markerscale':1.4})
ax.title.set_text("Personal Income by NJ County")

ax.margins(0.04)
```

```python
AHDataSummary = AHDataSummary.iloc[0:]
AHDataSummary.rename(columns={'Affordable Housing Municipal Status Report as of March 6, 2025.1': 'Name of Housing Development'}, inplace=True)
AHDataSummary.head()
```

```python
AHDataSummary.rename(columns={'Housing Name of Housing Development': 'Name of Housing Development'}, inplace=True)
AHDataSummary.head()
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, figsize=(6,10))
njCO_merged.plot(ax=ax, color='white', edgecolor='black',linewidth=1)
# Fixed: Added .plot to the DataFrame object
AHDataSummary.plot(column='Name of Housing Development',markersize=50,legend=True,cmap='RdYlGn_r',scheme='natural_breaks',k=5,ax=ax)


from matplotlib.lines import Line2D
point = Line2D([0], [0], label='Developments', marker='X', markersize=3,
         markeredgecolor='r', markerfacecolor='r', linestyle='')
plt.legend(handles=[point])
```

```python
# 1. Inspect columns to locate county and clean the dataset
# In AHDataSummary, Column 'Unnamed: 6' contains county names (e.g., 'Atlantic')

# Standardize county names to uppercase to match njCO_merged
AHDataSummary['County_Clean'] = AHDataSummary['Unnamed: 6'].astype(str).str.strip().str.upper()

# Group and count the number of developments in each county
developments_per_county = (
    AHDataSummary.groupby('County_Clean')
    .size()
    .reset_index(name='Development_Count')
)

# 2. Merge the development counts into our spatial GeoDataFrame
njCO_housing_map = njCO_merged.merge(
    developments_per_county,
    left_on='COUNTY',
    right_on='County_Clean',
    how='left'
)

# Fill NaN values with 0 for counties with no reported developments
njCO_housing_map['Development_Count'] = njCO_housing_map['Development_Count'].fillna(0)

# 3. Plot the map showing the concentration of affordable housing developments
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, figsize=(8, 12))

njCO_housing_map.plot(
    ax=ax,
    column='Development_Count',
    legend=True,
    cmap='RdYlGn_r',
    scheme='natural_breaks',
    k=5,
    edgecolor='black',
    linewidth=1,
    legend_kwds={
        'fmt': '{:,.0f}',
        'loc': 'lower right',
        'title': 'Number of Developments',
        'title_fontsize': 'medium',
        'fontsize': 'small'
    }
)

ax.set_title("Affordable Housing Developments by NJ County", fontsize=14)
ax.axis('off')
plt.show()
```
